```python
# ==========================================
# STEP 1: DATASET COLLECTION & PROBLEM STATEMENT
# ==========================================
# Dataset Name: Students Performance in Exams
# Source URL: https://www.kaggle.com/datasets/spscientist/students-performance-in-exams
# Problem Statement: Analyze how demographic factors, parental education, lunch type,
# and test preparation courses impact student performance across Math, Reading, and Writing.

# ==========================================
# STEP 2: PYTHON AND NUMPY OPERATIONS
# ==========================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn.preprocessing import LabelEncoder, StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Load dataset temporarily to extract numerical data for NumPy operations
temp_df = pd.read_csv('StudentsPerformance.csv')
math_scores = temp_df['math score'].to_numpy()
reading_scores = temp_df['reading score'].to_numpy()

# Array creation, indexing, slicing, and arithmetic operations
sample_arr = np.array([10, 20, 30, 40, 50])
print("First 5 Math Scores (Slicing):", math_scores[:5])
print("10th Student Math Score (Indexing):", math_scores[9])

total_two_subjects = math_scores + reading_scores  # Arithmetic operation
print("Combined Math & Reading (First 5):", total_two_subjects[:5])

# Statistical calculations using NumPy
print(f"Math Mean: {np.mean(math_scores):.2f}")
print(f"Math Median: {np.median(math_scores):.2f}")
print(f"Math Std Dev: {np.std(math_scores):.2f}")
print(f"Math Min: {np.min(math_scores)}, Max: {np.max(math_scores)}")
print(f"Math 25th, 50th, 75th Percentiles: {np.percentile(math_scores, [25, 50, 75])}")

# ==========================================
# STEP 3: PANDAS OPERATIONS
# ==========================================
df = pd.read_csv('StudentsPerformance.csv')

print("\n--- First 5 Records ---")
display(df.head())

print("\n--- Last 5 Records ---")
display(df.tail())

print(f"\nDataset Shape: {df.shape[0]} rows and {df.shape[1]} columns")
print("\n--- Column Names & Data Types ---")
print(df.dtypes)

print("\n--- Summary Statistics ---")
display(df.describe())

print("\n--- Missing Values ---")
print(df.isnull().sum())

print(f"\nDuplicate Records Count: {df.duplicated().sum()}")

# ==========================================
# STEP 4: DATA CLEANING AND PREPROCESSING
# ==========================================
df_before = df.copy()

# 1. Clean column names (replace spaces and slashes with underscores)
df.columns = [col.strip().lower().replace(' ', '_').replace('/', '_') for col in df.columns]

# 2. Handle duplicates (if any)
df.drop_duplicates(inplace=True)

# 3. Handle missing values (demonstration: fill numeric with median, categorical with mode)
for col in df.select_dtypes(include=np.number).columns:
    df[col].fillna(df[col].median(), inplace=True)
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

# 4. Ensure consistent data types
df['math_score'] = df['math_score'].astype(int)

# 5. Outlier detection using IQR on math_score
Q1 = df['math_score'].quantile(0.25)
Q3 = df['math_score'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['math_score'] < (Q1 - 1.5 * IQR)) | (df['math_score'] > (Q3 + 1.5 * IQR))]
print(f"\nNumber of Outliers detected in Math Score: {len(outliers)}")

# 6. Compare dataset before and after cleaning
print(f"Shape Before Cleaning: {df_before.shape} | Shape After Cleaning: {df.shape}")
print("Cleaned Columns:", df.columns.tolist())

# ==========================================
# STEP 5: EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================
print("\n--- Categorical Value Counts ---")
for col in ['gender', 'lunch', 'test_preparation_course']:
    print(f"\n{col.upper()}:\n", df[col].value_counts())

print("\n--- Correlation Matrix ---")
numeric_cols = df.select_dtypes(include=np.number)
display(numeric_cols.corr())

# Observations (Include in Markdown cell):
# 1. Reading and writing scores have a very strong positive correlation (~0.95).
# 2. Students who completed the test preparation course scored higher on average across all subjects.
# 3. Students with standard lunch performed significantly better than those with free/reduced lunch.
# 4. Female students scored higher in reading and writing, whereas male students scored slightly higher in math.
# 5. Math scores contain a few low-end outliers (scores below 30).

# ==========================================
# STEP 6: DATA VISUALIZATION USING MATPLOTLIB
# ==========================================
plt.figure(figsize=(16, 12))

# 1. Histogram
plt.subplot(2, 3, 1)
plt.hist(df['math_score'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')

# 2. Bar Chart
plt.subplot(2, 3, 2)
prep_means = df.groupby('test_preparation_course')['math_score'].mean()
plt.bar(prep_means.index, prep_means.values, color=['salmon', 'lightgreen'])
plt.title('Average Math Score by Test Prep')
plt.xlabel('Test Preparation Course')
plt.ylabel('Average Math Score')

# 3. Line Chart (Average score across sorted student index sample)
plt.subplot(2, 3, 3)
sample_sorted = df.sort_values(by='math_score').reset_index()
plt.plot(sample_sorted.index[:50], sample_sorted['math_score'][:50], label='Math', color='blue')
plt.plot(sample_sorted.index[:50], sample_sorted['reading_score'][:50], label='Reading', color='orange')
plt.title('Score Trend (First 50 Sorted Students)')
plt.xlabel('Student Index')
plt.ylabel('Score')
plt.legend()

# 4. Scatter Plot
plt.subplot(2, 3, 4)
plt.scatter(df['reading_score'], df['writing_score'], alpha=0.6, color='purple')
plt.title('Reading vs Writing Score')
plt.xlabel('Reading Score')
plt.ylabel('Writing Score')

# 5. Box Plot
plt.subplot(2, 3, 5)
plt.boxplot([df['math_score'], df['reading_score'], df['writing_score']], labels=['Math', 'Reading', 'Writing'])
plt.title('Box Plot of Exam Scores')
plt.ylabel('Score')

# 6. Correlation Heatmap
plt.subplot(2, 3, 6)
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')

plt.tight_layout()
plt.savefig('matplotlib_visualizations.png')
plt.show()

# ==========================================
# STEP 7: INTERACTIVE VISUALIZATION USING PLOTLY
# ==========================================
# 1. Interactive Histogram
fig1 = px.histogram(df, x='math_score', color='gender', barmode='overlay',
                    title='Interactive Histogram: Math Score Distribution by Gender',
                    labels={'math_score': 'Math Score', 'gender': 'Gender'})
fig1.show()

# 2. Interactive Scatter Plot
fig2 = px.scatter(df, x='reading_score', y='writing_score', color='test_preparation_course',
                  hover_data=['gender', 'math_score'],
                  title='Interactive Scatter Plot: Reading vs Writing Scores by Test Prep',
                  labels={'reading_score': 'Reading Score', 'writing_score': 'Writing Score'})
fig2.show()

# 3. Interactive Bar Chart
avg_scores = df.groupby('parental_level_of_education')[['math_score', 'reading_score', 'writing_score']].mean().reset_index()
fig3 = px.bar(avg_scores, x='parental_level_of_education', y=['math_score', 'reading_score', 'writing_score'],
              barmode='group', title='Interactive Bar Chart: Average Scores by Parental Education',
              labels={'value': 'Average Score', 'parental_level_of_education': 'Parental Education'})
fig3.show()

# ==========================================
# STEP 8: FEATURE ENGINEERING
# ==========================================
# 1. Create 3 meaningful new features
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['percentage'] = (df['total_score'] / 300) * 100
df['performance_grade'] = pd.cut(df['percentage'], bins=[0, 50, 70, 85, 100], labels=['D', 'C', 'B', 'A'])

# Note: Since this dataset has no date/time column, we demonstrate date extraction on a synthetic exam_date column
df['exam_date'] = pd.date_range(start='2026-01-01', periods=len(df), freq='H')
df['exam_month'] = df['exam_date'].dt.month
df['exam_day_of_week'] = df['exam_date'].dt.day_name()

# 2. Categorical Encoding
le = LabelEncoder()
df['gender_encoded'] = le.fit_transform(df['gender'])
df = pd.get_dummies(df, columns=['lunch', 'test_preparation_course'], drop_first=True)

# 3. Numerical Scaling
scaler = StandardScaler()
df[['math_scaled', 'reading_scaled', 'writing_scaled']] = scaler.fit_transform(
    df[['math_score', 'reading_score', 'writing_score']]
)

print("\n--- Engineered Dataset Sample ---")
display(df[['total_score', 'percentage', 'performance_grade', 'gender_encoded', 'math_scaled']].head())
```

### Step 5: Exploratory Data Analysis (EDA) — Key Observations
Based on the statistical summary, categorical distributions, and correlation analysis, here are 5 key observations:

1. **Strong Linear Correlation Between Language Scores:** Reading and writing scores show an exceptionally strong positive correlation ($r \approx 0.95$), indicating that students who excel in reading comprehension almost always perform equally well in written expression.
2. **Impact of Test Preparation Courses:** Students who completed the test preparation course achieved higher mean and median scores across all three subjects (Math, Reading, and Writing) compared to those who did not take the course.
3. **Socioeconomic & Nutritional Influence:** Students receiving a `standard` lunch scored significantly higher on average than students receiving a `free/reduced` lunch, suggesting a strong link between socioeconomic stability/nutrition and academic performance.
4. **Gender-Based Subject Trends:** Female students outperformed male students on average in Reading and Writing, whereas male students demonstrated a slightly higher average score in Mathematics.
5. **Presence of Low-End Outliers:** While most exam scores cluster between 60 and 80, the distribution is slightly left-skewed due to a small number of low-scoring outliers (below 30), particularly in Mathematics.

### Step 7: Interpretation of Interactive Plotly Visualizations

1. **Interactive Histogram (Math Score Distribution by Gender):**
   - Hovering over the overlapping bars reveals that the peak frequency for both genders lies between **60 and 75 marks**.
   - Male students have a slightly higher representation in the upper tail (85–100 marks), while extreme low scores (0–20) appear more prominently among a few outliers.

2. **Interactive Scatter Plot (Reading vs. Writing Scores by Test Prep):**
   - The scatter plot confirms a tight, upward-sloping linear trend between reading and writing scores.
   - Filtering or hovering over the legend shows that points representing students who **completed** the test preparation course are heavily concentrated in the top-right quadrant (high reading and high writing scores).

3. **Interactive Bar Chart (Average Scores by Parental Education):**
   - Grouping average subject scores by parental education level demonstrates a clear upward progression: students whose parents hold a **Master's or Bachelor's degree** consistently achieve the highest mean scores across all subjects, whereas students whose parents had `some high school` or `high school` education show lower averages.

   ### Step 8: Purpose and Importance of Engineered Features

1. **`total_score` & `percentage`:**
   - **Purpose:** Combines individual subject scores (`math_score`, `reading_score`, `writing_score`) into a single aggregate metric out of 300 and normalizes it to a 100-point scale.
   - **Importance:** Provides a holistic measure of overall academic achievement rather than evaluating subjects in isolation.

2. **`performance_grade` (A, B, C, D):**
   - **Purpose:** Bins continuous percentage values into discrete academic performance tiers.
   - **Importance:** Simplifies high-level reporting, enables cohort segmentation, and prepares the target variable for potential classification tasks.

3. **Date/Time Features (`exam_month`, `exam_day_of_week`):**
   - **Purpose:** Demonstrates temporal feature extraction by deriving month and weekday attributes from timestamp data.
   - **Importance:** In real-world academic datasets, temporal features help uncover seasonal or day-of-week performance patterns.

4. **Categorical Encoding (`LabelEncoder` & One-Hot Encoding):**
   - **Purpose:** Converts text categories (`gender`, `lunch`, `test_preparation_course`) into machine-readable binary/numeric formats.
   - **Importance:** Required by statistical and machine learning algorithms that cannot process raw string labels directly.

5. **Numerical Scaling (`StandardScaler`):**
   - **Purpose:** Standardizes subject scores to have a mean of $0$ and a standard deviation of $1$.
   - **Importance:** Ensures all numerical features contribute equally to distance-based algorithms without bias toward larger numerical ranges.

   ### Step 9: Analysis, Conclusions, and Limitations

#### Major Findings & Patterns
- Academic performance across subjects is interconnected, with reading and writing showing the strongest link ($r \approx 0.95$), and math correlating moderately ($r \approx 0.80$) with both.
- External support factors—specifically completing a test preparation course, having standard lunch access, and higher parental education—consistently shift score distributions upward.

#### Conclusions & Recommendations
1. **Expand Test Preparation Access:** Since completing the test preparation course yields measurable gains across all subjects, schools should make test-prep workshops mandatory or widely accessible, especially for underperforming cohorts.
2. **Targeted Support for Free/Reduced Lunch Students:** Educational institutions should pair nutritional support programs with academic mentoring to bridge the performance gap observed between `standard` and `free/reduced` lunch groups.
3. **Cross-Disciplinary Literacy Focus:** Because reading proficiency strongly drives writing and even supports math word-problem comprehension, early interventions in reading comprehension can create positive spillover effects across all subjects.

#### Dataset & Analysis Limitations
- **Synthetic/Anonymized Origin:** The dataset contains 1,000 records generated for educational analysis rather than longitudinal real-world school records, which limits real-world generalizability.
- **Missing Behavioral Variables:** Key factors such as daily study hours, attendance rates, teacher quality, and school funding are absent from the dataset.
- **Correlation vs. Causation:** Observed relationships (e.g., parental education vs. student grades) reflect statistical associations rather than direct causal proof.
